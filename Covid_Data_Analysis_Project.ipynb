```python
#active Cases
covid_df['Active_Cases'] = covid_df['Confirmed'] - (covid_df['Cured']+covid_df['Deaths'])
covid_df.tail()
```

```python
statewise = pd.pivot_table(covid_df,values=["Confirmed","Deaths","Cured"],index= "State/UnionTerritory",aggfunc=max)
```

```python
statewise["Recovery Rate"]=statewise["Cured"]*100/statewise["Confirmed"]
```

```python
statewise["Mortality Rate"]=statewise["Deaths"]*100/statewise["Confirmed"]
```

```python
statewise=statewise.sort_values(by="Confirmed",ascending=False)
```

```python
statewise.style.background_gradient(cmap = "cubehelix")
```

```python
#top 10 Active cases states
top_10_active_cases = covid_df.groupby(by='State/UnionTerritory').max()[['Active_Cases','Date']].sort_values(by = ['Active_Cases'],ascending = False)
```

```python
fig = plt.figure(figsize=(16, 9))
```

```python
#top 10 Active cases states
top_10_active_cases = covid_df.groupby(by='State/UnionTerritory').max()[['Active_Cases','Date']].sort_values(by = ['Active_Cases'],ascending = False)
fig = plt.figure(figsize=(10,5))
plt.title("Top 10 states with most active cases in India",size=20)
ax = sns.barplot(data=top_10_active_cases.iloc[:10],y="Active_Cases",x="State/UnionTerritory", palette="Spectral",linewidth=1,edgecolor = 'black' )
plt.xticks(rotation=45, ha='right')
plt.xlabel("States")
plt.ylabel("Total Active Cases")
plt.show()
```

```python
# Top States with highest deaths
top_10_deaths = covid_df.groupby(by = "State/UnionTerritory").max()[["Deaths","Date"]].sort_values(by=["Deaths"],ascending=False).reset_index()

fig = plt.figure(figsize=(18,5))
plt.title("Top 10 States with most deaths", size=25)
ax = sns.barplot(data=top_10_deaths.iloc[:10],y="Deaths",x="State/UnionTerritory",palette="viridis",linewidth=2,edgecolor="black")
plt.xlabel("States")
plt.ylabel("Total Deaths Cases")
plt.xticks(rotation=20, ha='right')
plt.show()
```

```python
vaccine = vaccine_df[vaccine_df.State!="India"]
vaccine
```

```python
vaccine.rename(columns = {"Total Individuals Vaccinated": "Total"},inplace = True)
vaccine.head()
```

```python
#most 5 vaccinated states
max_vac = vaccine.groupby('State')['Total'].sum().to_frame('Total')
max_vac = max_vac.sort_values('Total',ascending = False)[:5]
max_vac['Total'].plot(kind='line', figsize=(8, 4), title='Top 5 Vaccinated States in India')
plt.gca().spines[['top', 'right']].set_visible(False)
plt.show()
```
